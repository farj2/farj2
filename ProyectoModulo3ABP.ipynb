Lección 1: la librería NumPy
- Se crea un conjunto de datos usando NumPy y se usan operaciones básicas para la preparación inicial.

1. Para generar los datos se importa la librería numpy.
2. Se fija una semilla para reproducir los valores (elegí el valor 50 de forma aleatoria).
3. Se solicita crear un datasets de clientes y transacciones usando arrays, para esto el número de clientes (n_clientes) se fija en 20.
4. Los clientes tienen un ID (el número de cliente) y se les asigna una edad random dentro de un rango de 18 a 50 años.
5. Se asigna un valor de cantidad de transacciones a cada cliente dentro de un rango de 1 a 40.
6. Se genera un monto random de cantidad total de dinero gastado para cada cliente, para claridad del ejercicio se asumirá que está en pesos chilenos (CLP).
7. Luego se juntan todos los datos en un array.

```python
import numpy as np

np.random.seed(50)

n_clientes = 20

#De forma opcional, se les asigna un "ID" a los clientes, para mayor orden
clientes_id = np.arange(1, n_clientes + 1)
edades = np.random.randint(18, 50, n_clientes)
transacciones = np.random.randint(1, 41, n_clientes)
montos = np.random.randint(10000, 500000, n_clientes)

#Generación de matriz
data = np.array([clientes_id, edades, transacciones, montos]).T
```

Para satisfacer el uso de operaciones básicas se usa:
- sum(montos) para sumar los montos totales de los 20 clientes en conjunto.
- mean(montos) para obtener el promedio de los montos y mean(edades) para obtener el promedio de las edades de los 20 clientes.
- len(clientes_id) hace un conteo de la cantidad de clientes en base a sus ID.
- max(montos) y min(montos) para obtener el mayor y el menor de los montos de las transacciones.

```python
total_montos = np.sum(montos)

promedio_montos = np.mean(montos)
promedio_edad = np.mean(edades)

total_transacciones = np.sum(transacciones)

conteo_clientes = len(clientes_id)

max_gasto = np.max(montos)
min_gasto = np.min(montos)

print("Total montos:", total_montos)
print("Promedio montos:", promedio_montos)
print("Promedio edad:", promedio_edad)
print("Total transacciones:", total_transacciones)
print("Cantidad clientes:", conteo_clientes)
print("Mayor gasto:", max_gasto)
print("Menor gasto:", min_gasto)
```

Se guarda el dataset generado en un archivo .npy

```python
np.save("clientes_data.npy", data)
```

Lección 2: la librería Pandas
 Exploración y transformación de datos generados en Lección 1.

- Se carga el .npy de la Lección 1.

```python
import numpy as np
import pandas as pd

data=np.load("clientes_data.npy")

df = pd.DataFrame(data, columns=["cliente_id", "edad", "transacciones", "monto"])

print(df.head())
```

La exploración inicial cuenta con:
1. Visualización de primeras head() y últimas filas tail().
2. Obtención de estadísticas descriptivas describe() e info().
3. Aplicación de filtros condicionales.
 - Clientes con gastos superiores a los $300.000
 - Clientes con más de 15 transacciones
 - Clientes menores de 30 años y gastos superiores a $200.000
 - Gasto promedio por transacción


```python
print(df.head())
print(df.tail())
```

```python
print(df.describe())
print(df.info())
```

```python
clientes_altos = df[df["monto"] > 300000]
print(clientes_altos.head())
```

```python
clientes_frecuentes = df[df["transacciones"] > 10]
print(clientes_frecuentes.head())
```

```python
segmento = df[(df["edad"] < 30) & (df["monto"] > 200000)]
print(segmento.head())
```

```python
# Gasto promedio por transacción
df["gasto_promedio"] = df["monto"] / df["transacciones"]

print(df.head())
```

Para guardar el nuevo dataframe en formato CSV se usa df.to_csv()

```python
df.to_csv("clientes_preprocesados.csv", index=False)
```
