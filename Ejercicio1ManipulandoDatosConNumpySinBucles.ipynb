Ejercicio 1 Manipulando Datos Con NumPy Sin Bucles

```python
import numpy as np
valores=np.arange(100)
print(valores)
```

```python
matriz=valores.reshape([10,10])
print(matriz)
```

```python
zirtam=matriz.T
print(zirtam)
```

```python
zirtamm=zirtam*2
print(zirtamm)
```

```python
diagonal=np.diag(zirtamm)
print(diagonal)
```

```python
identitty=np.eye(10)
print(identitty)
```

```python
ultimatriz=zirtamm@identitty
print(ultimatriz)
```
