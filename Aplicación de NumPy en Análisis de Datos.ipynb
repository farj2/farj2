Para optimizar la manipulación de datos se pretende aplicar NumPy, así mejorando eficiencia en cálculos financieros y facilitando la toma de decisiones.

Algunas operaciones que se deben abarcar son:
- Creación y manipulación de arrays multidimensionales.

- Indexación y selección eficiente de elementos dentro de los arrays.
- Optimización del rendimiento computacional mediante el uso de broadcasting y operaciones vectorizadas.

El primer paso es crear la matriz de datos utilizando la librería NumPy.
Se fija una semilla (42) para posterior reproducción y obtener igual datos.
Los datos se generan con un randomm entre valores de 100 y 500 para obtener un buen rango de variabilidad de datos.

```python
import numpy as np

# Fijar semilla para reproducibilidad del ejercicio
np.random.seed(42)

# Generar datos simulados (los precios variarán entre 100 y 500)
precios = np.random.uniform(100, 500, size=(5, 5))

print("Matriz de precios:\n", precios)
```

Para obtener el promedio de cada acción durante los 5 días se utiliza el axis=1 y así recorre cada acción.
Al igual para obtener los máximos y mínimos de cada acción se usa la instrucción axis=1.

```python
promedio = np.mean(precios, axis=1)
maximo = np.max(precios, axis=1)
minimo = np.min(precios, axis=1)

print("Promedio por acción:", promedio)
print("Máximo por acción:", maximo)
print("Mínimo por acción:", minimo)
```

Para obtener la variación porcentual diaria se usa la fórmula de ((Valor Final-Valor Inicial)/Valor Inicial)x100 que se ve expresada como la diferencia del primer y último día de cada acción y en valor porcentual de 0-1 en valores positivos o negativos según corresponda.

```python
variacion = np.diff(precios, axis=1) / precios[:, :-1]

print("Variación porcentual diaria:\n", variacion)
```

La función np.log permite obtener los algoritmos de cada valor de los 5 días de cada acción.

```python
log_precios = np.log(precios)
print(log_precios)
```

Con la funcióm np.exp se obtiene el valor exponencial de cada cada, en este caso se están dividiendo en 100 para que la expresión en output no sean tan grandes y evitar extra consumo de memoria.

```python
exp_precios = np.exp(precios / 100)  # escalado para evitar overflow
print(exp_precios)
```

Los valores de las acciones son normalizados con la función de mínimo y máximo de cada precio. 

```python
min_vals = precios.min(axis=1, keepdims=True)
max_vals = precios.max(axis=1, keepdims=True)

normalizados = (precios - min_vals) / (max_vals - min_vals)

print("Datos normalizados:\n", normalizados)
```

Para obtener el valor específico de algún día para una acción se usa la indexación avanzada, donde se puede escoger en eje X la acción y en eje Y el día.
En este caso, como ejemplo, se usa [2,3] que quiere decir la acción 3 en el día 4 (recordar que en Python las matrices y valores comienzan en 0).

```python
precio_especifico = precios[2, 3]
print("Precio específico:", precio_especifico)
```

Otro método de indexación avanzada que permite ver varias acciones durante varios días o incluso la misma acción en distintos días:

```python
acciones = [0, 2, 4]
dias = [1, 3, 4]

seleccion = precios[acciones, dias]
print("Selección avanzada:", seleccion)
```

Un ejemplo de Broadcasting es este que permite ajustar los valores de todas las acciones en todos los días a una inflación general, como ejemplo se utiliza una inflación de 2% (lo que es lo mismo que representarlo de la forma 1.02).

```python
inflacion = 1.02
precios_ajustados = precios * inflacion
print(precios_ajustados)
```

En el siguiente ejemplo de Broadcasting, se le resta el promedio de los 5 días de cada acción al valor de cada acción de todos los días.

```python
precios_centrados = precios - promedio[:, np.newaxis]
print(precios_centrados)
```
